### static prompt

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

llm = ChatOpenAI(model="gpt-6-luna")

prompts = [
    {"role":"system","content":"I am javascript developer"},
    {"role":"user","content":"what is this keyword"}
]

res = llm.invoke(prompts)
print(res.content)
```

```python

from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

load_dotenv()

llm = ChatOpenAI(model="gpt-6-luna") 

prompts =ChatPromptTemplate.from_messages([
    {"role":"system", "content":"i am {language} developer"},
    {"role":"user","content":"{query}"}
])
finalePrompt = prompts.format_messages(language="python", query="what is map filter")
res = llm.invoke(finalePrompt)
print(res.content)
```
