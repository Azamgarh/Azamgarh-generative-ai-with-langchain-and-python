### create basic Agent

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent


llm = ChatGroq(model="openai/gpt-oss-20b")


@tool
def add_number(a:int, b:int):
    """ creating two number and thier sum"""
    return a+b


@tool
def multiply_number(a:int, b:int):
    """ creating two number and thier multification"""
    return a*b


agent  = create_agent(
    model=llm,
    tools=[add_number, multiply_number],
    system_prompt="you are a math teacher always use for calculation"
)

response = agent.invoke({"messages":[{"role":"user","content":"what is  5+15 *5 ?"}]})

print(response["messages"][-1].content)












```
