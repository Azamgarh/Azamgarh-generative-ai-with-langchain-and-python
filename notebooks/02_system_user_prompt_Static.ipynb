### System and User Prompt

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()

llm = ChatOpenAI(model="gpt-6-luna")
# using tupple
prompts  = [
    ("system","you are a angular developer"),
    ( "user","explain service" )
]
res = llm.invoke(prompts)
print(res.content)




```

### static prompt using dictionary 

```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv()
llm = ChatOpenAI(model='gpt-6-luna')

prompts = [
    {'role':'system', 'content':"you are a angular developer"},
    {'role':'user', 'content':'Explain modules'}
]
res = llm.invoke(prompts)
print(res.content)

```
